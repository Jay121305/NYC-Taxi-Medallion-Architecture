```python
%sql
CREATE OR REPLACE TABLE workspace.nyc_taxi.demo_customers (
    customer_id INT,
    name STRING,
    city STRING
);
```

```python
%sql
INSERT INTO workspace.nyc_taxi.demo_customers VALUES
(1, 'Alice', 'London'),
(2, 'Bob', 'Manchester'),
(3, 'Charlie', 'Birmingham');
```

```python
%sql
SELECT *
FROM workspace.nyc_taxi.demo_customers;
```

```python
%sql
CREATE OR REPLACE TABLE workspace.nyc_taxi.demo_updates (
    customer_id INT,
    name STRING,
    city STRING
);
```

```python
%sql
INSERT INTO workspace.nyc_taxi.demo_updates VALUES
(2, 'Bob', 'Leeds'),
(4, 'David', 'Liverpool');
```

```python
%sql
MERGE INTO workspace.nyc_taxi.demo_customers AS target
USING workspace.nyc_taxi.demo_updates AS source
ON target.customer_id = source.customer_id

WHEN MATCHED THEN
UPDATE SET
    target.name = source.name,
    target.city = source.city

WHEN NOT MATCHED THEN
INSERT (
    customer_id,
    name,
    city
)
VALUES (
    source.customer_id,
    source.name,
    source.city
);
```

```python
%sql
SELECT *
FROM workspace.nyc_taxi.demo_customers
ORDER BY customer_id;
```

```python
%sql
DESCRIBE HISTORY workspace.nyc_taxi.demo_customers;
```
