```python
%sql
SELECT
    DATE(tpep_pickup_datetime) AS trip_date,
    ROUND(SUM(total_amount),2) AS revenue,
    COUNT(*) AS trips
FROM workspace.nyc_taxi.silver_trips
GROUP BY DATE(tpep_pickup_datetime)
ORDER BY revenue DESC;
```

```python
%sql
SELECT
    Zone,
    Borough,
    Trips,
    ROUND(Revenue,2) AS Revenue
FROM workspace.nyc_taxi.gold_zone_summary
ORDER BY Trips DESC
LIMIT 10;
```

```python
%sql
SELECT
    payment_type,
    COUNT(*) AS trips,
    ROUND(SUM(total_amount),2) AS revenue
FROM workspace.nyc_taxi.silver_trips
GROUP BY payment_type
ORDER BY trips DESC;
```

```python
%sql
SELECT
    DATE(tpep_pickup_datetime) AS trip_date,

    ROUND(SUM(total_amount),2) AS revenue,

    RANK() OVER(
        ORDER BY SUM(total_amount) DESC
    ) AS revenue_rank

FROM workspace.nyc_taxi.silver_trips

GROUP BY DATE(tpep_pickup_datetime);
```

```python
%sql
SELECT
    VendorID,

    ROUND(SUM(total_amount),2) AS revenue,

    DENSE_RANK() OVER(
        ORDER BY SUM(total_amount) DESC
    ) AS rank

FROM workspace.nyc_taxi.silver_trips

GROUP BY VendorID;
```

```python
%sql
SELECT

DATE(tpep_pickup_datetime) AS trip_date,

SUM(total_amount) AS daily_revenue,

SUM(
SUM(total_amount)
)
OVER(
ORDER BY DATE(tpep_pickup_datetime)
)
AS cumulative_revenue

FROM workspace.nyc_taxi.silver_trips

GROUP BY DATE(tpep_pickup_datetime)

ORDER BY trip_date;
```

```python
%sql
SELECT

VendorID,

ROUND(
AVG(total_amount),2
) AS avg_revenue

FROM workspace.nyc_taxi.silver_trips

GROUP BY VendorID;
```

```python
%sql
SELECT

HOUR(tpep_pickup_datetime) AS hour,

COUNT(*) AS trips

FROM workspace.nyc_taxi.silver_trips

GROUP BY HOUR(tpep_pickup_datetime)

ORDER BY trips DESC;
```

```python
%sql
CREATE OR REPLACE VIEW workspace.nyc_taxi.daily_revenue_view AS

SELECT

DATE(tpep_pickup_datetime) AS trip_date,

ROUND(
SUM(total_amount),2
) AS revenue

FROM workspace.nyc_taxi.silver_trips

GROUP BY DATE(tpep_pickup_datetime);
```

```python
%sql
SELECT *
FROM workspace.nyc_taxi.daily_revenue_view;
```

```python
%sql
WITH daily AS (
    SELECT
        CAST(tpep_pickup_datetime AS DATE) AS trip_date,
        SUM(total_amount) AS revenue
    FROM workspace.nyc_taxi.silver_trips
    GROUP BY CAST(tpep_pickup_datetime AS DATE)
)

SELECT
    trip_date,
    ROUND(revenue, 2) AS revenue
FROM daily
WHERE revenue > 100000
ORDER BY revenue DESC;
```
