```python
%sql
SELECT
FORMAT_NUMBER(COUNT(*), 0) AS Total_Trips
FROM workspace.nyc_taxi.silver_trips;
```

```python
%sql
SELECT
CONCAT('$', FORMAT_NUMBER(SUM(total_amount), 2)) AS Total_Revenue
FROM workspace.nyc_taxi.silver_trips;
```

```python
%sql
SELECT
CONCAT('$', FORMAT_NUMBER(AVG(fare_amount), 2)) AS Average_Fare
FROM workspace.nyc_taxi.silver_trips;
```

```python
%sql
SELECT
CONCAT(FORMAT_NUMBER(AVG(trip_distance), 2), ' miles') AS Avg_Distance
FROM workspace.nyc_taxi.silver_trips;
```

```python
%sql
SELECT
trip_date,
total_revenue
FROM workspace.nyc_taxi.gold_daily_summary
WHERE trip_date BETWEEN DATE('2024-01-01')
AND DATE('2024-03-31')
ORDER BY trip_date;
```

```python
%sql
SELECT
pickup_hour,
total_trips
FROM workspace.nyc_taxi.gold_hourly_summary
ORDER BY pickup_hour;
```

```python
%sql
SELECT
CASE payment_type
    WHEN 0 THEN 'Flex Fare / Unknown'
    WHEN 1 THEN 'Credit Card'
    WHEN 2 THEN 'Cash'
    WHEN 3 THEN 'No Charge'
    WHEN 4 THEN 'Dispute'
    WHEN 5 THEN 'Unknown'
    WHEN 6 THEN 'Voided Trip'
    ELSE 'Other'
END AS Payment_Method,

trips

FROM workspace.nyc_taxi.gold_payment_summary

ORDER BY trips DESC;
```

```python
%sql
SELECT

CONCAT('Vendor ', VendorID) AS Vendor,

total_trips

FROM workspace.nyc_taxi.gold_vendor_summary

ORDER BY total_trips DESC;
```

```python
%sql
SELECT
Zone,
Trips
FROM workspace.nyc_taxi.gold_zone_summary
ORDER BY Trips DESC
LIMIT 10;
```
